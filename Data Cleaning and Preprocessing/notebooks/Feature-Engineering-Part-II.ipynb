### Binning (Discretization)

Binning is the process of converting continuous numerical data into categorical groups or bins. Instead of using exact values like 500, 800, 1000, 1200, 1500, we group them into categories.

**Example:**
- 0 - 600 sq ft → Small House
- 600 - 1200 sq ft → Medium House
- 1200+ sq ft → Large House

This is called discretization.


### Why We Need Binning?

1. **Reduce Noise** - Small numerical changes don't always matter.
   - Example: 1001 sq ft vs 1003 sq ft is almost same. Binning treats them equally.

2. **Cleaner Patterns** - Models sometimes learn better with ranges instead of exact values.

3. **Business Logic** - Makes it easier for non-technical people to understand.


### Types of Binning

**1. Manual Binning (Domain-Based)**
You decide the ranges based on your knowledge or business rules.

**2. Equal-Width Binning**
Divides data into bins of equal size.

**3. Equal-Frequency Binning**
Divides data into bins with equal number of observations.


### Manual Binning Example

Let's create bins for the 'area' feature:

| Area Range | Category |
|------------|----------|
| 0 - 3000 | Very Small |
| 3000 - 6000 | Small |
| 6000 - 9000 | Medium |
| 9000 - 12000 | Large |
| 12000+ | Very Large |



## Step by Step Impletation.


# Step 1: Load cleaned house data.

```python
import pandas as pd
df=pd.read_csv("../Data/Housing_cleaned.csv")
df.head()
```

# Step 2: Basic analysis.

```python
df.info()
df.describe()
```

# Step 3: Binning on Area

```python
# Define bins
bins=[0,2500,4000,6000,8000,10000]
labels=["very_small","small","medium","large","very_large"]
df["area_bin"]=pd.cut(df["area"],bins=bins,labels=labels)
df.head(10)
```

# Step 4: Convert price to Lakhs



```python
df["price"]=df["price"]/100000
df.head()
```

# Step 5: Binning on Price

```python
bins=[0, 25,40,60,80,100]
labels=["low-Budget","medium-budget","high-budget","luxury-budget","ultra-luxury-budget"]
df["price_bin"]=pd.cut(df["price"],bins=bins,labels=labels)
df.head(10)
```

# Step 6: Convert Bins into Feature vector.(One-Hot-Encoding)



### Encoding

Encoding is the process of converting categorical values into numerical values so that machine learning models can train on them.


### One-Hot Encoding

Each category becomes a separate binary feature (0 or 1). Only one feature is active at a time.

**Example:**

Houses = ["Small", "Medium", "Large"]

**One-Hot Encoding:**

| House   | Small | Medium | Large |
|---------|-------|--------|-------|
| Small   | 1     | 0      | 0     |
| Medium  | 0     | 1      | 0     |
| Large   | 0     | 0      | 1     |

### Why Use One-Hot Encoding?

1. **No Order Assumption** - Does not imply any ranking between categories
2. **Works with All Models** - Linear, Tree-based, Neural Networks
3. **Simple to Implement** - Easy to understand and use
4. **Preserves Information** - No loss of categorical information


# Step 6: Encode on Area bin and Price Bin.

```python
df["mainroad"]=df["mainroad"].map({"yes":1,"no":0})
df_encoded=pd.get_dummies(df,columns=["area_bin","price_bin","furnishingstatus"],drop_first=True)
df_encoded.head()
```

# Step 7: Convert True/False to 1/0.

```python
df_encoded = df_encoded.replace({True: 1, False: 0})
df_encoded.head()
```
