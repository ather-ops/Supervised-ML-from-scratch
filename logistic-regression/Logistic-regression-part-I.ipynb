### **Logistic Regression :** 
Logistic Regression is a supervised machine learning algorithm primarily used for **classification problems**. It predicts the probability that an input belongs to a particular class.

Unlike Linear Regression, which predicts continuous numerical values, Logistic Regression is commonly used for binary classification, where the target has two possible classes.

### Examples
- Email classification: Spam (1) or Not Spam (0)
- Student prediction: Pass (1) or Fail (0)
- Medical classification: Positive (1) or Negative (0)
- Customer prediction: Will Purchase (1) or Will Not Purchase (0)

## Linear Regression vs Logistic Regression

| Feature | Linear Regression | Logistic Regression |
|---|---|---|
| Main purpose | Predict continuous values | Predict class probabilities |
| Output | Any real number | Probability between 0 and 1 |
| Common task | Price prediction | Binary classification |
| Example | Predict house price | Predict whether a customer buys |

### Why Is Logistic Regression Needed?

A linear equation can produce predictions below 0 or above 1, which are not valid probabilities.

Logistic Regression uses the sigmoid function to transform the model's raw output into a probability between 0 and 1.

# Logistic Regression 

## 1. Linear Combination

Logistic Regression first calculates a linear combination of the input features:

\[
z = wx + b
\]

Where:
- \(x\) = input feature
- \(w\) = model weight
- \(b\) = model bias
- \(z\) = raw model output

## 2. Sigmoid Function

The sigmoid function converts the raw output into a probability:

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

The sigmoid function produces values between 0 and 1.

- If \(z\) is a large negative number, the output approaches 0.
- If \(z = 0\), the output is 0.5.
- If \(z\) is a large positive number, the output approaches 1.

## 3. Classification Threshold

A probability is converted into a predicted class using a threshold.


y = 
\begin{cases}
1, & \text{if } \sigma(z) \geq 0.5 \\
0, & \text{if } \sigma(z) < 0.5
\end{cases}


The threshold of 0.5 is a common default and can be adjusted for different problems.

## 4. How Does the Model Learn?

During training, Logistic Regression adjusts its weights and bias to improve its predictions.

## Classification Threshold

Logistic Regression produces probabilities, but classification tasks generally require discrete class labels.

We use a threshold to convert each probability into a predicted class.

Using the default threshold of 0.5:

$$
\hat{y} =
\begin{cases}
1, & \text{if } p \geq 0.5 \\
0, & \text{if } p < 0.5
\end{cases}
$$

Where:

- \(p\) = predicted probability
- \(y\) = predicted class

### Example

| Predicted probability | Predicted class |
|---:|---:|
| 0.15 | 0 |
| 0.35 | 0 |
| 0.50 | 1 |
| 0.78 | 1 |
| 0.95 | 1 |

The threshold of 0.5 is a common default. Depending on the problem, we can adjust it to make classification more or less sensitive to class 1.

## How Does Logistic Regression Learn?

Initially, the model does not know the correct values of its weight \(w\) and bias \(b\). It starts with initial parameter values and improves them during training.

The learning process involves the following steps:

1. **Initialize parameters:** Assign initial values to the weight and bias.
2. **Predict probabilities:** Calculate probabilities using the sigmoid function.
3. **Calculate loss:** Measure how far the predictions are from the actual labels using Binary Cross-Entropy.
4. **Calculate gradients:** Determine how the weight and bias should change to reduce the loss.
5. **Update parameters:** Use Gradient Descent to adjust the weight and bias.
6. **Repeat:** Continue the process for multiple training iterations, called epochs.

The goal is to find parameter values that minimize the loss on the training data.

### Training workflow

$$
\text{Input} \rightarrow \text{Probability} \rightarrow \text{Loss}
$$

$$
\rightarrow \text{Gradients} \rightarrow \text{Parameter Updates}
$$

### Sigmoid function

```python

import numpy as np
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
print("Done")
```

### Prdicting probability

```python
def predict_probability(x,w,b):
    z = w*x+b
    pred_prob = sigmoid(z)
    return pred_prob

print("Done")
```

### Applying threshold 0.5 for predict class

```python
def predict_class(x,w,b):
    p = predict_probability(x,w,b)
    return (p >= 0.5).astype(bool)
print("Done")
```

### Preparing the Dataset

```python
X = np.array([1,2,3,4,5,6,7]) # Study hours
y = np.array([0,0,0,0,1,1,1])  # pass fail

print(f"Shape of X :{X.shape} and Shape of y :{y.shape}")
print(f"X represents study hours and y represents pass/fail!")
```

### Binary Cross-Entropy Loss

```python
def log_loss(y,p):
    loss = -(y * np.log(p) + (1-y) * np.log(1-p))
    return loss
print(log_loss(1, 0.8))
print(log_loss(0, 0.2))
print(log_loss(1, 0.3))
print(log_loss(0, 0.8))
```

### Average loss across multiple samples

```python
def average_loss(y, p):
    n = len(y)
    loss = -(1/n) * np.sum(
        y * np.log(p) + (1-y) * np.log(1-p)
    )
    return loss
print("Done")
```

### Gradient descent Train function

```python
def train(x,y):
    w = 0.0
    b = 0.0
    lr = 0.1
    n = len(x)
    epochs = 100
    loss_history = []
    for epoch in range(epochs):
        p = predict_probability(x,w,b)


        loss = -(1/n) * np.sum(
            y * np.log(p) + (1-y) * np.log(1-p)
            )

        dw = (1/n)*np.sum((p-y)*x)
        db = (1/n)*np.sum(p-y)

        w -= lr*dw
        b -= lr*db

        loss_history.append(loss)

    return w,b,loss_history
print("Done")
```

```python
# test
w,b,loss_history = train(X,y)
z = predict_probability(X,w,b)
pred_class = predict_class(X,w,b)

print("Predict Claases :",pred_class)
```

### Test on new unseen data

```python
X_new = np.array([2.5,4.5,6.5,1.5])
new_prob = predict_probability(X_new,w,b)
new_pred_class = predict_class(X_new,w,b)

for hours,prob, pred in zip(X_new,new_prob,new_pred_class):
    result = "pass" if pred else "fail"
    print(f"Study hours : {hours:.1f} | "
          f"Prbability of passing : {(prob)*100:.2f}% | "
          f"prediction : {result}")
```

### Saving the Trained Model

After training, we can save the learned weight and bias so that we can reuse the model without training it again.

For this simple Logistic Regression model, the learned parameters are sufficient to reproduce its predictions, provided we retain the same model and preprocessing logic.

We will use NumPy's `savez()` function to store the parameters in a file.

```python
import os
model_dir = os.path.join(os.getcwd(), "model")
os.makedirs(model_dir, exist_ok=True)
model_path = os.path.join(model_dir, "logistic_regression_part_I_model.npz")
np.savez(model_path, w=w, b=b)
print("Model folder exists:", os.path.isdir(model_dir))
print("Model file exists:", os.path.isfile(model_path))
```
