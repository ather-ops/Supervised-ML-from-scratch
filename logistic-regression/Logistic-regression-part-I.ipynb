### **Logistic Regression :** 
Logistic Regression is a supervised machine learning algorithm primarily used for **classification problems**. It predicts the probability that an input belongs to a particular class.

Unlike Linear Regression, which predicts continuous numerical values, Logistic Regression is commonly used for binary classification, where the target has two possible classes.

### Examples
- Email classification: Spam (1) or Not Spam (0)
- Student prediction: Pass (1) or Fail (0)
- Medical classification: Positive (1) or Negative (0)
- Customer prediction: Will Purchase (1) or Will Not Purchase (0)

## Linear Regression vs Logistic Regression

| Feature | Linear Regression | Logistic Regression |
|---|---|---|
| Main purpose | Predict continuous values | Predict class probabilities |
| Output | Any real number | Probability between 0 and 1 |
| Common task | Price prediction | Binary classification |
| Example | Predict house price | Predict whether a customer buys |

### Why Is Logistic Regression Needed?

A linear equation can produce predictions below 0 or above 1, which are not valid probabilities.

Logistic Regression uses the sigmoid function to transform the model's raw output into a probability between 0 and 1.

# Logistic Regression 

## 1. Linear Combination

Logistic Regression first calculates a linear combination of the input features:

\[
z = wx + b
\]

Where:
- \(x\) = input feature
- \(w\) = model weight
- \(b\) = model bias
- \(z\) = raw model output

## 2. Sigmoid Function

The sigmoid function converts the raw output into a probability:

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

The sigmoid function produces values between 0 and 1.

- If \(z\) is a large negative number, the output approaches 0.
- If \(z = 0\), the output is 0.5.
- If \(z\) is a large positive number, the output approaches 1.

## 3. Classification Threshold

A probability is converted into a predicted class using a threshold.


y = 
\begin{cases}
1, & \text{if } \sigma(z) \geq 0.5 \\
0, & \text{if } \sigma(z) < 0.5
\end{cases}


The threshold of 0.5 is a common default and can be adjusted for different problems.

## 4. How Does the Model Learn?

During training, Logistic Regression adjusts its weights and bias to improve its predictions.

### Hands on Impletation

### Sigmoid function

```python

import numpy as np
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


print(sigmoid(0))
print(sigmoid(2))
print(sigmoid(-2))
```
